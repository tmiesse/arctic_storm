```python

def compute_keep_idx2(
    sample_path,
    min_depth,
    max_depth,
    bbox=None,
):
    """
    Return indices of nodes whose depth is between min_depth and max_depth,
    optionally also lying within a bounding box in x/y.

    Parameters
    ----------
    sample_path : str or Path
        Path to a CF‑ified fort.63 file.
    min_depth, max_depth : float
        Depth range (inclusive) in meters.
    bbox : tuple of floats (min_x, max_x, min_y, max_y), optional
        If given, further restrict to nodes with:
           min_x <= x <= max_x  AND  min_y <= y <= max_y.

    Returns
    -------
    keep_idx : np.ndarray of int
        Indices of the `node` dimension satisfying your criteria.
    """
    ds0 = xr.open_dataset(sample_path, engine="netcdf4", mask_and_scale=False)
    
    # 1) depth mask
    depth = ds0.depth.values
    fv    = ds0.depth.encoding.get("_FillValue", None)
    mask = np.ones_like(depth, dtype=bool)
    if fv is not None:
        mask &= (depth != fv)
    mask &= (depth >= min_depth)
    mask &= (depth <= max_depth)

    # 2) optional bbox mask
    if bbox is not None:
        min_x, max_x, min_y, max_y = bbox
        x = ds0.x.values
        y = ds0.y.values
        mask &= (x >= min_x) & (x <= max_x)
        mask &= (y >= min_y) & (y <= max_y)

    ds0.close()
    return np.where(mask)[0]
```

```python
import ruptures as rpt

# 1) Load your monthly max dataset

# dims = ('time','node')

# 2) Extract the times and the DataArray we’ll analyze
times = ds_monthly['time'].values     # array of numpy.datetime64
zeta  = ds_monthly['zeta']            # (time, node)

def detect_cp_monthly(y, times, n_bkps=1):
    # 1) Mask NaNs
    idx = np.where(~np.isnan(y))[0]
    if idx.size < 6:
        return np.datetime64("NaT")

    arr = y[idx].astype(float)
    # 2) Try the rupture; on *any* error bail out
    try:
        algo = rpt.Binseg(model="rbf").fit(arr.reshape(-1,1))
        bkps = algo.predict(n_bkps=n_bkps)
        cp   = bkps[0]
    except Exception:
        return np.datetime64("NaT")

    # 3) If break is interior, return its timestamp
    if 0 < cp < len(arr):
        return times[idx[cp]]
    return np.datetime64("NaT")

# 3) Vectorize over the node dimension
da_inf = xr.apply_ufunc(
    detect_cp_monthly,
    zeta,
    input_core_dims=[['time']],
    output_core_dims=[[]],          # scalar per node
    vectorize=True,
    kwargs={'times': times},
    dask='parallelized',
    output_dtypes=['datetime64[ns]'])
```

```python
importance_method = "pearson"  # options: "pearson", "spearman", or "partial"
from scipy.stats import spearmanr, rankdata, pearsonr
from sklearn.linear_model import LinearRegression
sea_list = ['Bering Sea', 'Chukchi Sea', 'Beaufort Sea']
out_dir = '/scratch/tmiesse/project/figures'


nnode     = ds_inf_mon.dims['node']
times     = ds_monthly['time'].values
inf_dates = ds_inf_mon['inflection_date'].values


pre_frac_ice_pearson  = np.full(nnode, np.nan)
pre_frac_wind_pearson = np.full(nnode, np.nan)
pre_frac_pres_pearson = np.full(nnode, np.nan)
post_frac_ice_pearson  = np.full(nnode, np.nan)
post_frac_wind_pearson = np.full(nnode, np.nan)
post_frac_pres_pearson = np.full(nnode, np.nan)
# Spearman
pre_frac_ice_spear  = np.full(nnode, np.nan)
pre_frac_wind_spear = np.full(nnode, np.nan)
pre_frac_pres_spear = np.full(nnode, np.nan)
post_frac_ice_spear  = np.full(nnode, np.nan)
post_frac_wind_spear = np.full(nnode, np.nan)
post_frac_pres_spear = np.full(nnode, np.nan)
# Partial
pre_frac_ice_partial  = np.full(nnode, np.nan)
pre_frac_wind_partial = np.full(nnode, np.nan)
pre_frac_pres_partial = np.full(nnode, np.nan)
post_frac_ice_partial  = np.full(nnode, np.nan)
post_frac_wind_partial = np.full(nnode, np.nan)
post_frac_pres_partial = np.full(nnode, np.nan)

# 2) Pull arrays
zeta_arr = ds_monthly['zeta'].values          # (time, node)
ice_arr  = ds_monthly['sea_ice'].values
wind_arr = ds_monthly['wind_speed'].values
pres_arr = ds_monthly['pressure'].values

# 3) Helper functions
from scipy.stats import spearmanr, kendalltau

def distance_correlation(x, y):
    x = np.asarray(x); y = np.asarray(y)
    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]; y = y[mask]
    n = x.size
    if n < 2:
        return 0.0
    def _dm(a): return np.abs(a[:, None] - a[None, :])
    A = _dm(x); B = _dm(y)
    def _center(D):
        row_mean = D.mean(axis=1, keepdims=True)
        col_mean = D.mean(axis=0, keepdims=True)
        total_mean = D.mean()
        return D - row_mean - col_mean + total_mean
    A_c = _center(A)
    B_c = _center(B)
    dcov2 = np.mean(A_c * B_c)
    dvar_x = np.mean(A_c * A_c)
    dvar_y = np.mean(B_c * B_c)
    if dvar_x <= 0 or dvar_y <= 0:
        return 0.0
    return np.sqrt(dcov2) / np.sqrt(np.sqrt(dvar_x * dvar_y))

def compute_mixed_importance(mask, i, tau_thresh=0.3, p_thresh=0.05):
    if mask.sum() < 4:
        return (np.nan, np.nan, np.nan)
    z = zeta_arr[mask, i]
    ice = ice_arr[mask, i]
    wind = wind_arr[mask, i]
    pres = pres_arr[mask, i]

    def score(driver):
        valid = np.isfinite(z) & np.isfinite(driver)
        if valid.sum() < 4:
            return 0.0
        z_v = z[valid]
        d_v = driver[valid]
        tau, p_tau = kendalltau(z_v, d_v)
        if (not np.isnan(tau)) and (abs(tau) >= tau_thresh) and (p_tau < p_thresh):
            rho, _ = spearmanr(z_v, d_v)
            return abs(rho) if not np.isnan(rho) else 0.0
        else:
            return distance_correlation(z_v, d_v)

    r_ice  = score(ice)
    r_wind = score(wind)
    r_pres = score(pres)
    total = r_ice + r_wind + r_pres
    if total > 0:
        return (r_ice/total, r_wind/total, r_pres/total)
    else:
        return (0.0, 0.0, 0.0)
    
def compute_pearson_importance(mask, i):
    if mask.sum() < 4:
        return (np.nan, np.nan, np.nan)
    rs = []
    for arr in (ice_arr, wind_arr, pres_arr):
        x = zeta_arr[mask, i]
        y = arr[mask, i]
        valid = np.isfinite(x) & np.isfinite(y)
        if valid.sum() < 4:
            rs.append(0.0)
        else:
            r, _ = pearsonr(x[valid], y[valid])
            rs.append(abs(r) if not np.isnan(r) else 0.0)
    total = sum(rs)
    if total > 0:
        return tuple(r/total for r in rs)
    else:
        return (0.0, 0.0, 0.0)

def compute_spearman_importance(mask, i):
    if mask.sum() < 4:
        return (np.nan, np.nan, np.nan)
    rs = []
    for arr in (ice_arr, wind_arr, pres_arr):
        x = zeta_arr[mask, i]
        y = arr[mask, i]
        valid = np.isfinite(x) & np.isfinite(y)
        if valid.sum() < 4:
            rs.append(0.0)
        else:
            rho, _ = spearmanr(x[valid], y[valid])
            rs.append(abs(rho) if not np.isnan(rho) else 0.0)
    total = sum(rs)
    if total > 0:
        return tuple(r/total for r in rs)
    else:
        return (0.0, 0.0, 0.0)

def partial_spearman(x, y, controls):
    # only keep samples where x, y, and all controls are finite
    mask = np.isfinite(x) & np.isfinite(y)
    for c in controls:
        mask &= np.isfinite(c)
    if mask.sum() < 4:
        return 0.0
    # rank-transform the filtered data
    xr = rankdata(x[mask])
    yr = rankdata(y[mask])
    cr = np.vstack([rankdata(c[mask]) for c in controls]).T  # shape (n_samples, n_controls)

    # regress ranked zeta on other drivers and get residuals
    lm_z = LinearRegression().fit(cr, xr)
    resid_z = xr - lm_z.predict(cr)

    # regress ranked driver on other drivers and get residuals
    lm_y = LinearRegression().fit(cr, yr)
    resid_y = yr - lm_y.predict(cr)

    # Pearson on residuals approximates partial Spearman
    corr = np.corrcoef(resid_z, resid_y)[0, 1]
    if np.isnan(corr):
        return 0.0
    return abs(corr)

def compute_partial_importance(mask, i):
    if mask.sum() < 4:
        return (np.nan, np.nan, np.nan)
    x = zeta_arr[mask, i]
    ice = ice_arr[mask, i]
    wind = wind_arr[mask, i]
    pres = pres_arr[mask, i]
    r_ice  = partial_spearman(x, ice,  [wind, pres])
    r_wind = partial_spearman(x, wind, [ice, pres])
    r_pres = partial_spearman(x, pres,  [ice, wind])
    total = r_ice + r_wind + r_pres
    if total > 0:
        return (r_ice / total, r_wind / total, r_pres / total)
    else:
        return (0.0, 0.0, 0.0)

# 4) Loop over nodes to compute all three normalized importances
pre_frac_ice  = np.full((nnode,), np.nan)
pre_frac_wind = np.full((nnode,), np.nan)
pre_frac_pres = np.full((nnode,), np.nan)
post_frac_ice  = np.full((nnode,), np.nan)
post_frac_wind = np.full((nnode,), np.nan)
post_frac_pres = np.full((nnode,), np.nan)

for i in range(nnode):
    cut = inf_dates[i]
    if np.isnat(cut):
        continue

    pre_mask  = times < cut
    post_mask = times >= cut

    pre_frac_ice[i], pre_frac_wind[i], pre_frac_pres[i] = compute_mixed_importance(pre_mask, i)
    post_frac_ice[i], post_frac_wind[i], post_frac_pres[i] = compute_mixed_importance(post_mask, i)

# 5) Build ds_imp including all methods
ds_imp = xr.Dataset({
    'pre_frac_ice':  (('node',), pre_frac_ice),
    'pre_frac_wind': (('node',), pre_frac_wind),
    'pre_frac_pres': (('node',), pre_frac_pres),
    'post_frac_ice':  (('node',), post_frac_ice),
    'post_frac_wind': (('node',), post_frac_wind),
    'post_frac_pres': (('node',), post_frac_pres),
}, coords={
    'node': ds_inf_mon['node'],
    'x': ('node', ds_inf_mon['x'].values),
    'y': ('node', ds_inf_mon['y'].values),
    'inflection_date': ('node', inf_dates)
})
```

```python
from scipy.stats import kendalltau, spearmanr, pearsonr
from sklearn.isotonic import IsotonicRegression
from sklearn.metrics import mean_squared_error
import numpy as np

def test_monotonicity(zeta, driver,
                      min_samples=10,
                      tau_thresh=0.3,
                      p_thresh=0.05):
    # 1) filter finite
    mask = np.isfinite(zeta) & np.isfinite(driver)
    n = mask.sum()
    if n < min_samples:
        return {
            "n": n,
            "is_monotonic": False,
            "kendall_tau": np.nan,
            "kendall_p": np.nan,
            "spearman_rho": np.nan,
            "pearson_r": np.nan,
            "iso_mse": np.nan,
            "lin_mse": np.nan,
        }

    z = zeta[mask]
    d = driver[mask]

    # 2) rank/order metrics
    try:
        tau, p_tau = kendalltau(z, d)
    except Exception:
        tau, p_tau = np.nan, np.nan
    try:
        rho, _ = spearmanr(z, d)
    except Exception:
        rho = np.nan
    try:
        r, _ = pearsonr(z, d)
    except Exception:
        r = np.nan

    # 3) isotonic vs linear fit MSE
    mse_iso = np.nan
    mse_lin = np.nan
    # need at least some variability
    if len(np.unique(d)) >= 2 and len(np.unique(z)) >= 2:
        # sort by driver for smooth plotting/comparison
        order = np.argsort(d)
        d_sorted = d[order]
        z_sorted = z[order]

        try:
            ir = IsotonicRegression(out_of_bounds="clip")
            z_iso = ir.fit_transform(d_sorted, z_sorted)
            mse_iso = mean_squared_error(z_sorted, z_iso)
        except Exception:
            mse_iso = np.nan

        # linear fit fallback: use np.linalg.lstsq explicitly to avoid polyfit SVD issues
        try:
            A = np.vstack([d_sorted, np.ones_like(d_sorted)]).T  # design matrix
            coeffs, _, _, _ = np.linalg.lstsq(A, z_sorted, rcond=None)
            z_lin = A @ coeffs
            mse_lin = mean_squared_error(z_sorted, z_lin)
        except Exception:
            mse_lin = np.nan

    is_monotonic = False
    if (not np.isnan(tau)) and (abs(tau) >= tau_thresh) and (p_tau < p_thresh):
        is_monotonic = True

    return {
        "n": n,
        "is_monotonic": is_monotonic,
        "kendall_tau": tau,
        "kendall_p": p_tau,
        "spearman_rho": rho,
        "pearson_r": r,
        "iso_mse": mse_iso,
        "lin_mse": mse_lin,
    }
```

```python
import numpy as np
import xarray as xr
from collections import Counter
import matplotlib.pyplot as plt
import ternary
from matplotlib.cm import get_cmap, ScalarMappable
from matplotlib.colors import BoundaryNorm
from scipy.stats import spearmanr, kendalltau, pearsonr

# --- user config ----------------------------------------------------
importance_method = "mixed"  # "pearson", "spearman", "partial", or "mixed"
res = 100  # ternary resolution
outpath = f'/scratch/tmiesse/project/figures/ternary_{importance_method}.png'
# -------------------------------------------------------------------

# ---- assumptions: ds_inf_mon and ds_monthly are already loaded ----
# you need these arrays for the mixed method
inf_dates = ds_inf_mon['inflection_date'].values
times = ds_monthly['time'].values
zeta_arr = ds_monthly['zeta'].values          # (time, node)
ice_arr  = ds_monthly['sea_ice'].values
wind_arr = ds_monthly['wind_speed'].values
pres_arr = ds_monthly['pressure'].values
nnode = ds_inf_mon.dims['node']

# --- helper for mixed importance ----------------------------------
def distance_correlation(x, y):
    x = np.asarray(x); y = np.asarray(y)
    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]; y = y[mask]
    n = x.size
    if n < 2:
        return 0.0
    def _dm(a): return np.abs(a[:, None] - a[None, :])
    A = _dm(x); B = _dm(y)
    def _center(D):
        row_mean = D.mean(axis=1, keepdims=True)
        col_mean = D.mean(axis=0, keepdims=True)
        total_mean = D.mean()
        return D - row_mean - col_mean + total_mean
    A_c = _center(A)
    B_c = _center(B)
    dcov2 = np.mean(A_c * B_c)
    dvar_x = np.mean(A_c * A_c)
    dvar_y = np.mean(B_c * B_c)
    if dvar_x <= 0 or dvar_y <= 0 or dcov2 < 0:
        return 0.0
    return np.sqrt(dcov2) / np.sqrt(np.sqrt(dvar_x * dvar_y))

def compute_mixed_importance(mask, i, tau_thresh=0.3, p_thresh=0.05):
    if mask.sum() < 4:
        return (np.nan, np.nan, np.nan)
    z = zeta_arr[mask, i]
    ice = ice_arr[mask, i]
    wind = wind_arr[mask, i]
    pres = pres_arr[mask, i]

    def score(driver):
        valid = np.isfinite(z) & np.isfinite(driver)
        if valid.sum() < 4:
            return 0.0
        z_v = z[valid]
        d_v = driver[valid]
        tau, p_tau = kendalltau(z_v, d_v)
        if (not np.isnan(tau)) and (abs(tau) >= tau_thresh) and (p_tau < p_thresh):
            rho, _ = spearmanr(z_v, d_v)
            return abs(rho) if not np.isnan(rho) else 0.0
        else:
            return distance_correlation(z_v, d_v)

    r_ice  = score(ice)
    r_wind = score(wind)
    r_pres = score(pres)
    total = r_ice + r_wind + r_pres
    if total > 0:
        return (r_ice/total, r_wind/total, r_pres/total)
    else:
        return (0.0, 0.0, 0.0)

# --- stacking logic ------------------------------------------------
def stack_frac_arrays(ds_imp, method):
    if method == "pearson":
        pre_i = ds_imp['pre_frac_ice_pearson'].values
        pre_w = ds_imp['pre_frac_wind_pearson'].values
        pre_p = ds_imp['pre_frac_pres_pearson'].values
        post_i = ds_imp['post_frac_ice_pearson'].values
        post_w = ds_imp['post_frac_wind_pearson'].values
        post_p = ds_imp['post_frac_pres_pearson'].values
    elif method == "spearman":
        pre_i = ds_imp['pre_frac_ice_spearman'].values
        pre_w = ds_imp['pre_frac_wind_spearman'].values
        pre_p = ds_imp['pre_frac_pres_spearman'].values
        post_i = ds_imp['post_frac_ice_spearman'].values
        post_w = ds_imp['post_frac_wind_spearman'].values
        post_p = ds_imp['post_frac_pres_spearman'].values
    elif method == "partial":
        pre_i = ds_imp['pre_frac_ice_partial'].values
        pre_w = ds_imp['pre_frac_wind_partial'].values
        pre_p = ds_imp['pre_frac_pres_partial'].values
        post_i = ds_imp['post_frac_ice_partial'].values
        post_w = ds_imp['post_frac_wind_partial'].values
        post_p = ds_imp['post_frac_pres_partial'].values
    else:
        raise ValueError(f"Unknown importance_method '{method}'")
    frac_pre_raw = np.vstack([pre_i, pre_w, pre_p])   # shape (3, N)
    frac_post_raw = np.vstack([post_i, post_w, post_p])
    return frac_pre_raw, frac_post_raw

if importance_method == "mixed":
    pre_frac_ice  = np.full((nnode,), np.nan)
    pre_frac_wind = np.full((nnode,), np.nan)
    pre_frac_pres = np.full((nnode,), np.nan)
    post_frac_ice  = np.full((nnode,), np.nan)
    post_frac_wind = np.full((nnode,), np.nan)
    post_frac_pres = np.full((nnode,), np.nan)

    for i in range(nnode):
        cut = inf_dates[i]
        if np.isnat(cut):
            continue
        pre_mask  = times < cut
        post_mask = times >= cut
        pre_frac_ice[i], pre_frac_wind[i], pre_frac_pres[i] = compute_mixed_importance(pre_mask, i)
        post_frac_ice[i], post_frac_wind[i], post_frac_pres[i] = compute_mixed_importance(post_mask, i)

    frac_pre_raw = np.vstack([pre_frac_ice, pre_frac_wind, pre_frac_pres])
    frac_post_raw = np.vstack([post_frac_ice, post_frac_wind, post_frac_pres])
else:
    # expects ds_imp to exist with the required variables for non-mixed methods
    frac_pre_raw, frac_post_raw = stack_frac_arrays(ds_imp, importance_method)
```
